```python
import pandas as pd #import library 

d= pd.read_csv("loan_data.csv")
d
```

```python
nd = d.select_dtypes(["int64","float64"])
nd
```

```python
nd.corr() #identify correlation between 
```

```python
x=nd[["person_income"]]# input
y=nd["loan_status"]# output
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.25, random_state=0)
x_train
```

```python
from sklearn.linear_model import LogisticRegression # download the logictic regression class

model = LogisticRegression()
model.fit(x_train,y_train)# fit model with the training input data to get the trainin output data
```

```python
y_prediction = model.predict(x_test)# passing the test data
y_prediction
```

```python
#find accuracy using r2 score
```

```python
from sklearn.metrics import r2_score
r2_score(y_test,y_prediction)
```

```python
# our r2 score is a negative number. this is incorrect. this meand that we must use a DIFFERENT accuracy scoerer, that is specific to the LOGISTIC REGRESSION
# for finsing the accuracy of the LOGICTIC REGRESSION, we use the ACCURACY SCORE, not the r2 score
```

```python
from sklearn.metrics import accuracy_score #NOTE: this is a form of EDA process
accuracy_score(y_test,y_prediction)
```
