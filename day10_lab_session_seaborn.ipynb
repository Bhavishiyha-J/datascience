```python
import seaborn as sns
import matplotlib.pyplot as plt

a= sns.load_dataset("tips")
a
```

```python
#task 1
```

```python
sns.histplot(a["total_bill"], kde=True)
plt.title("Total bill")
plt.xlabel("bill")
plt.ylabel("count")
plt.show()
```

```python
#task 2
```

```python
sns.boxplot(x="day",y="total_bill",data=a,hue="day")
plt.title("Tips across the days")
plt.show()
```

```python
#task 3
```

```python
sns.violinplot(x="day",y="tip",data=a, hue="day",split=True)
plt.title("distribution of tips for each day")
plt.show()
```

```python
#task 4
```

```python
sns.barplot(x="day",y="total_bill",data=a,hue="day")
plt.title("average bills per day")
plt.xlabel("days")
plt.ylabel("average bills")
plt.show()
```

```python
#5
```

```python
sns.countplot(x="size",data=a)
plt.title("avg number of customers each day")
plt.xlabel("days")
plt.ylabel("size")
plt.show()
```

```python
#task 6
```

```python
sns.scatterplot(x="total_bill",y="tip",data=a,hue="sex")
plt.title("relationship between total bill and tips for each sex")
plt.xlabel("total bill")
plt.ylabel("tip")
plt.grid()
plt.show()
```

```python
#task 7
```

```python
sns.scatterplot(x="total_bill",y="tip",data=a,hue="smoker", style="smoker")
plt.title("relationship between total bill and tips for each sex")
plt.xlabel("total bill")
plt.ylabel("tip")
plt.grid()
plt.show()
```
