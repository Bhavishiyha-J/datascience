```python
from sklearn.model_selection import train_test_split
x=df[["Fare"]]
y=df["Pclass"]
x_test,x_train,y_test,y_train=train_test_split(x,y,train_size=0.3,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
lr= LinearRegression()
lr.fit(x_train,y_train)
```

```python
y_prediction=lr.predict(x_test)
y_prediction
```

```python
lr.coef_#find gradient
```

```python
lr.intercept_#function for y intercept
```

```python
from sklearn.metrics import r2_score # r2 is used to find accuracy
r2_score(y_test,y_prediction) # relates how close our training data is in comparison to our testing data
```
