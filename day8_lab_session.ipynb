```python
sales={"Month":["Jan","Feb","March","Apr","May","Jun","Jul","Aug","Sept","Oct","Nov","Dec"],
      "sold":[37,15,9,4,3,11,12,7,2,7,42,47],
      "expenses":[47,42,7,2,7,12,11,3,4,9,15,37]}
df_sales=pd.DataFrame(sales)
df_sales
#df_sales.groupby("Month")[["sold","expenses"]].plot(kind="line")
plt.plot(df_sales["Month"], df_sales["sold"], color= "red", marker='.',label = "sales",linestyle = 'dotted')
plt.plot(df_sales["Month"], df_sales["expenses"], color= "blue", marker='.', label= "expenses", ls= 'dashed')
plt.title("Sold vs expenses per each month")
plt.legend()

plt.xlabel("Month")
plt.ylabel("sales vs expenses per 1000")
plt.show()
```

```python
sales={"Month":["Jan","Feb","March","Apr","May","Jun","Jul","Aug","Sept","Oct","Nov","Dec"],
      "sold":[37,15,9,4,3,11,12,7,2,7,42,47],
      "expenses":[47,42,7,2,7,12,11,3,4,9,15,37]}
df_sales=pd.DataFrame(sales)
df_sales
bars=plt.bar(df_sales["Month"], df_sales["sold"])#put in variable so can be called later so the numerical value for no of sales is shown
plt.title("Monthly sales")
plt.xlabel("Months")
plt.ylabel("Sales")          # new function of .bar_label so we can label each bar.
                             # functions: ,padding=n (adds space in between the bar and the text we are about to add) 
                            #             ,fmt= (formats the text by adding stuff like dollar sign or decimal places of percentages) 
                             #            ,fontsize=n (changes font size)
                            #             ,color='acolor'(changes font color)
plt.bar_label(bars,padding=2)# calling the variable bars as a positional argument so the tru value of no of sales is printed above

plt.xticks(rotation= 45)
plt.show()
```

```python
sales={"Month":["Jan","Feb","March","Apr","May","Jun","Jul","Aug","Sept","Oct","Nov","Dec"],
      "sold":[37,15,9,4,3,11,12,7,2,7,42,47],
      "expenses":[47,42,7,2,7,12,11,3,4,9,15,37]}
df_sales=pd.DataFrame(sales)
df_sales
bar_h=plt.barh(df_sales["Month"], df_sales["expenses"])# using the .barh() function instead of the normal .bar() function so bar is printed horizontally
plt.title("Monthly expenses")
plt.xlabel("Months")
plt.ylabel("expenses")
plt.bar_label(bar_h,padding=3)# calling the variable bar_h as a positional argument so the tru value of no of sales is printed above
                               # new function of .bar_label() so we can label each bar
plt.show()
```

```python
sales={"Month":["Jan","Feb","March","Apr","May","Jun","Jul","Aug","Sept","Oct","Nov","Dec"],
      "sold": [28, 25, 32, 38, 45, 52, 58, 54, 42, 48, 65, 80],
       "expenses": [20, 19, 22, 25, 30, 34, 38, 36, 29, 31, 40, 50],
       "profit": [8, 6, 10, 13, 15, 18, 20, 18, 13, 17, 25, 30]}
df_sales=pd.DataFrame(sales)
df_sales
plt.plot(df_sales["Month"],df_sales["profit"], marker='.')
max_index= df_sales["profit"].idxmax()# .idxmax() allows us to find th index postition of the highest profit
max_month= df_sales.loc[max_index, "Month"] #loc uses the index we just worked out to fetch the exact month and profit value
max_profit = df_sales.loc[max_index,"profit"]
plt.plot(max_month,max_profit,marker='.', color ='red')#plotting the max
plt.title("Profit")
plt.xlabel("months")
plt.ylabel("years")
plt.grid()
plt.annotate(
    'Max Profit: 27 (Nov)',          # The text you want to display
    xy=(max_month, max_profit),       # Point A: Where the arrow points TO (the peak)
  #  xytext=(max_index - 2, max_profit - 5), # Point B: Where the text box sits
   # arrowprops=dict(facecolor='red')  # Draws the red arrow connecting Point B to Point A
)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
sales={"Month":["Jan","Feb","March","Apr","May","Jun","Jul","Aug","Sept","Oct","Nov","Dec"],
      "sold": [28, 25, 32, 38, 45, 52, 58, 54, 42, 48, 65, 80],
       "expenses": [20, 19, 22, 25, 30, 34, 38, 36, 29, 31, 40, 50],
       "profit": [8, 6, 10, 13, 15, 18, 20, 18, 13, 17, 25, 30]}
df_sales=pd.DataFrame(sales)
df_sales
df_sales.set_index("Month")["sold"].plot(kind="pie", autopct="%1.1f%%")# 4. Set "Month" as the index so month names are used as labels,
                                                                #    select the "sold" column, and plot it as a pie chart with percentage values
plt.show()
```

```python
plt.scatter(df_sales["sold"],df_sales["profit"])# 1. Create a scatter plot with "sold" on the X-axis and "profit" on the Y-axis
plt.grid()# 2. Add background gridlines to make coordinates easier to read
plt.xlabel("sales")# 3. Label the horizontal axis (X-axis) as "sales"
plt.ylabel("profit")# 4. Label the vertical axis (Y-axis) as "profit"
plt.show()# 5. Display the final scatter plot window
```
