```python
import pandas as pd

h= pd.read_csv ("Housing.csv")
h
```

```python
#EDA PROCESSING
```

```python
h.columns[h.isnull().any()] #check for any null values
```

```python
h.info # checking for any object values, as machine can't process objects.
```

```python
h1=pd.get_dummies(h,columns=["mainroad"],drop_first=True) # this creates a new column that converts the object into boolean, and then removes the original column, ALL IN ONE
h1
```

```python
#LABEL ENCODING
```

```python
from sklearn.preprocessing import LabelEncoder 
le= LabelEncoder() #create an object from the class
h1["guestroom"]=le.fit_transform(h1["guestroom"]) #fit_tranform to change the data type from one to another. from object into a label. it is giving the object/value a category
h1
```

```python
x = h1[["area","bedrooms","bathrooms"]] #inputs
y= h1["price"] #oupt
```

```python
from sklearn.model_selection import train_test_split #this function will take some data for training/practise, the rest will be for testing if its accurate prediction

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=0) #must use the order x_train,x_test,y_train,y_test. CANNOT CHANGE
x_train
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x_train,y_train)
```

```python
y_prediction = model.predict(x_test) # all the preduction for all the
y_prediction
```

```python
from sklearn.metrics import r2_score

r2_score(y_test, y_prediction)
```

```python
# this r2 score is too low and innaccurate. we must use FINE TUNING
```
