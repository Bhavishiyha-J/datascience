```python
import pandas as pd
d= {"Name":["Hi","Hello","Bye","BuBye","Yo"],
   "Salary":[18000,20000,22000,25000,30000]}
d
```

```python
df=pd.DataFrame(d)
df
```

```python
import matplotlib.pyplot as plt
plt.bar(d["Name"],d["Salary"], color=["red"])
plt.title("People salary")
plt.xlabel("name")
plt.ylabel("salary")
plt.show()
```
