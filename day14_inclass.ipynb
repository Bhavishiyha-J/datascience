```python
import numpy as np
normal_dist = np.random.normal(loc=50,scale=3,size=100)
normal_dist
```

```python
ns=np.median(normal_dist)
ns
```

```python
n=np.array([1000,2000,3000])
na=np.concatenate([normal_dist,n])
na
#array with
```

```python
np.mean(na)
```

```python
np.median(na)
```

```python
import pandas as pd #data frame 
df = pd.DataFrame({"Normal": normal_dist})
df
```

```python
plt.figure(figsize=(10,7))#height and width
plt.subplot(1,2,1)
sb.histplot(normal_dist,kde= True)#for plotting the line
plt.title("Normal dis")
plt.subplot(1,2,2)
sb.histplot(na,kde=True)
plt.title("Outliers")
```
