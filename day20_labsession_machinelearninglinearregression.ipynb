```python
d["Age"] = d["Age"].fillna(d["Age"].mean())
d["Age"] 
```

```python
e=pd.get_dummies(d["Sex"])
e
```

```python
d=pd.concat([d,e], axis = 1)
d
```

```python
d=d.drop(["Sex"],axis=1)
d
```

```python
from sklearn.model_selection import train_test_split
x=d[["Age"]]
y=d["Survived"] 
x_test,x_train,y_test,y_train = train_test_split(x,y, train_size=0.9,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression 
lr= LinearRegression() 
lr.fit(x_train,y_train)
```

```python
y_prediction=lr.predict(x_test)
y_prediction
```
